```python
# Step 3: Text Cleaning for Issue Descriptions
# Clean all issue descriptions by applying:
# ➢ Remove punctuation (.,!?-)
# ➢ Convert multiple spaces → single space
# ➢ Remove leading/trailing spaces
# ➢ Convert text to lowercase
# ➢ Replace slang/shorthand (example: "ok" → "okay")


def clean_text(text):
    for char in ".,!?-":
        text = text.replace(char, " ")

    text = " ".join(text.split())
    text = text.strip()
    text = text.lower()
    words = text.split()
    for i in range(len(words)):
        if words[i] == "ok":
            words[i] = "okay"

    text = " ".join(words)
    return text

for i in range(len(ticket_data['Issue_Description'])):
    ticket_data['Issue_Description'][i] = clean_text(
        ticket_data['Issue_Description'][i])

print(ticket_data)
```

```python
# Step 4: Keyword-Based Issue Insights
def count_tickets_with_word(word):
    count = 0

    for issue in ticket_data['Issue_Description']:
        if word.lower() in issue.lower().split():
            count += 1

    return count



print("Tickets containing 'poor':", count_tickets_with_word("poor"))
print("Tickets containing 'good':", count_tickets_with_word("good"))
print("Tickets containing 'slow':", count_tickets_with_word("slow"))
print("Tickets containing 'excellent':", count_tickets_with_word("excellent"))
```

```python
# 3. Find the Ticket With the Longest Issue Description
max_words = 0
max_index = 0

for i in range(len(ticket_data['Issue_Description'])):
    word_count = len(ticket_data['Issue_Description'][i].split())

    if word_count > max_words:
        max_words = word_count
        max_index = i

print("Ticket Number :", ticket_data['Ticket_No'][max_index])
print("Customer Name :", ticket_data['Customer_Name'][max_index])
print("Issue         :", ticket_data['Issue_Description'][max_index])
print("Word Count    :", max_words)
```
